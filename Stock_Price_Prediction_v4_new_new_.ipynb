# **Function to Set Random Seeds for Reproducibility**

## **Overview**

In the context of machine learning and data analysis, ensuring reproducibility of results is a critical aspect of research. Random seed initialization plays a key role in this process, as it helps in controlling the randomness of algorithms, especially those involving stochastic processes such as data shuffling, initialization of model parameters, or optimization procedures. The function `set_random_seeds` is designed to configure random seeds for several libraries, ensuring consistent and repeatable results across different runs of the same code.

## **Function Overview**

The primary goal of the `set_random_seeds` function is to set random seeds for various libraries, including `numpy`, Python’s built-in `random` module, TensorFlow, and XGBoost. By setting a fixed seed value, the function ensures that the results of any stochastic processes (such as random number generation or model initialization) remain consistent across multiple executions of the same code. This consistency is essential for debugging, testing, and comparing different models or algorithms in a controlled manner.

### **Key Components**

1. **Numpy Random Number Generator (RNG)**:
   - The function sets the seed for `numpy` using `np.random.seed(seed_value)`. This ensures that any random operations performed by `numpy`, such as generating random arrays or sampling data, produce the same results every time the code is executed.

2. **Python's Random RNG**:
   - Python's built-in `random` module is also configured with the same seed value using `random.seed(seed_value)`. This step is essential when using Python's native random functions, ensuring consistent behavior across functions that depend on random number generation.

3. **TensorFlow RNG**:
   - TensorFlow's random seed is set using `tf.random.set_seed(seed_value)`, which ensures that TensorFlow’s internal random operations, such as the initialization of weights in neural networks or data shuffling during training, remain reproducible. This is particularly important for deep learning models, where the initial weights can significantly impact the training process.

4. **XGBoost Verbosity and Seed**:
   - The verbosity of XGBoost is configured to 0 using `xgb.set_config(verbosity=0)` to suppress unnecessary output, ensuring a cleaner execution process. Additionally, the seed for XGBoost is set using the parameter `'seed': seed_value` to control the randomness in the model's training process, leading to consistent model performance and results across multiple runs.

## **Workflow**

1. **Initialization**:
   - The function is called at the beginning of the program with the default seed value of `42`. This choice of seed value is common in the field of machine learning and ensures consistency in results.
   
2. **Setting the Seeds**:
   - Each library’s random number generator is set to the specified seed value. This ensures that any random operation within `numpy`, Python’s `random`, TensorFlow, and XGBoost follows a deterministic pattern based on the initial seed.
   
3. **Suppression of XGBoost Verbosity**:
   - To prevent unnecessary logging during model training, the verbosity level of XGBoost is set to zero, ensuring a streamlined and silent execution, which is particularly useful during extensive experiments.

## **Conclusion**

By configuring random seeds across multiple libraries, the `set_random_seeds` function guarantees reproducibility in experiments involving stochastic processes. This approach is essential for ensuring that machine learning models and data processing workflows yield consistent results, which is crucial in both academic research and practical applications. The function also enhances the debugging and testing phases by eliminating the variability introduced by randomness, thereby promoting more reliable comparisons between models and experiments.


```python
# Function to set random seeds for reproducibility
def set_random_seeds(seed_value=42):
    """
    Configure random seed for reproducibility.
    Args:
    - seed_value: Seed for RNGs (default: 42).
    Returns:
    - None
    """
    # Set seed for numpy RNG
    np.random.seed(seed_value)

    # Set seed for Python's random RNG
    random.seed(seed_value)

    # Set seed for TensorFlow RNG
    tf.random.set_seed(seed_value)

    # Configure XGBoost verbosity to 0 (silent)
    xgb.set_config(verbosity=0)

    # Set seed for XGBoost parameters
    xgparam = {'seed': seed_value}

# Initialise random seed for consistency
set_random_seeds()
```

# **Stock Data Fetching and Processing Workflow**

## **Overview**

In this section, I focus on the collection and processing of stock market data from multiple sources. The primary objective is to automate the process of downloading stock data for specific tickers over defined time periods, process this data by selecting only the relevant information, and merge the results into a consolidated dataset for further analysis. The workflow involves fetching stock data, processing it, and ensuring the data is stored in a structured format for easy accessibility and analysis. This is an essential step in preparing the data for machine learning models and statistical analysis.

## **Function Overview**

This project consists of several key functions designed to streamline the process of downloading and processing stock data:

1. **`fetch_data(ticker, start, end, output_dir)`**:
   - This function fetches stock data for a specific ticker between the specified start and end dates. If the data already exists in the given directory, it loads the data from a CSV file; otherwise, it downloads the data using the `yfinance` library and saves it in CSV format for future use. The function returns the downloaded data as a pandas DataFrame and the number of rows in the dataset.

2. **`process_data(df, ticker)`**:
   - After fetching the raw stock data, this function processes the data by retaining only the "Adjusted Close" prices, which reflect corporate actions such as dividends and stock splits. The function renames the "Adjusted Close" column to the stock ticker symbol and returns the processed DataFrame, along with the number of rows after processing.

3. **`download_and_process_data(tickers, start, end, out_dir)`**:
   - This function downloads and processes stock data for a list of tickers over a given time period. It iterates through each ticker, fetches the raw data, processes it, and merges the processed data into a single DataFrame. Additionally, it tracks the number of data points before and after processing for each ticker, storing this information in a dictionary for future reference.

## **Workflow**

1. **Fetching and Processing Data**:
   - The code begins by fetching stock market data for specific tickers using the `yfinance` library, which provides access to historical financial data. The data is fetched for three different groups of tickers: training, testing, and extra datasets. Each ticker’s data is downloaded within a specific date range, saved to a CSV file, and then processed by retaining only the relevant columns (Date and Adjusted Close).

2. **Merging Data**:
   - Once the data for each ticker is processed, it is merged into a combined DataFrame based on the common `Date` column. This ensures that the final dataset contains time-aligned stock prices for multiple tickers, allowing for cross-comparison and analysis across different stocks.

3. **Tracking Data Points**:
   - During the downloading and processing stages, the code keeps track of the number of rows (or data points) before and after processing. This information is stored in a dictionary for each ticker, providing insights into how much data was available and how much was used after processing.

4. **Saving and Managing Data**:
   - The code ensures that the data is saved in structured directories for each dataset (training, testing, and extra data), facilitating easy access for subsequent tasks such as model training and evaluation. By organizing the data in this manner, it becomes easier to manage, update, and analyze as new data becomes available.

## **Dataset Details**

The tickers and date ranges for each dataset are as follows:

1. **Training Dataset**:
   - Tickers: ['LLOY.L', 'SHEL.L', 'AZN.L', 'RIO.L', 'TSCO.L', 'VOD.L', 'BA.L', 'LAND.L', 'REL.L', 'SGE.L']
   - Date Range: January 1, 2014 to December 31, 2023
   - Directory: `train_data_`

2. **Test Dataset**:
   - Tickers: ['BARC.L', 'DGE.L', 'BP.L', 'GSK.L', 'GLEN.L', 'OCDO.L', 'PRU.L', 'RR.L', 'BLND.L', 'ULVR.L']
   - Date Range: January 1, 2024 to March 31, 2024
   - Directory: `test_data_`

3. **Extra Dataset**:
   - Tickers: ['LGEN.L', 'IMB.L', 'SSE.L', 'SN.L', 'AAL.L', 'SBRY.L', 'WOOD.L', 'WEIR.L', 'SGRO.L', 'EXPN.L']
   - Date Range: January 1, 2024 to June 30, 2024
   - Directory: `extra_data_`

## **Conclusion**

This approach efficiently automates the collection and processing of stock market data, providing a well-structured dataset ready for analysis. The code ensures reproducibility by saving the data in CSV format and facilitates further analysis by organizing the data into easy-to-access directories. The tracking of data points before and after processing provides insights into the completeness of the datasets, ensuring transparency in the data preparation workflow. This setup is essential for building robust predictive models and performing in-depth financial analysis.


```python
def fetch_data(ticker, start, end, output_dir):
    """
    Fetch and save stock data for a ticker.
    Args:
    - ticker (str): Stock ticker symbol.
    - start (str): Start date (YYYY-MM-DD).
    - end (str): End date (YYYY-MM-DD).
    - output_dir (str): Directory to save data.
    Returns:
    - pd.DataFrame: Loaded data for ticker.
    - int: Number of rows in the downloaded data.
    """
    # Ensure output directory exists
    os.makedirs(output_dir, exist_ok=True)

    # Define file path for the data
    file_path = os.path.join(output_dir, f"{ticker}.csv")

    # Download data
    if not os.path.exists(file_path):
        data = yf.download(ticker, start=start, end=end)
        data.reset_index(inplace=True)  # Ensure 'Date' is a column
        data.to_csv(file_path, index=False)  # Save without index
    else:
        data = pd.read_csv(file_path)

    # Return the data as DataFrame and its row count
    return data, data.shape[0]

# Function to process data
def process_data(df, ticker):
    """
    Process data: keep 'Adj Close', rename.
    Args:
    - df (pd.DataFrame): Stock data.
    - ticker (str): Stock ticker symbol.
    Returns:
    - pd.DataFrame: Processed data with 'Adj Close'.
    - int: Number of rows after processing.
    """
    # Convert 'Date' to datetime format
    df['Date'] = pd.to_datetime(df['Date'])

    # Keep only 'Date' and 'Adj Close' columns
    df = df[['Date', 'Adj Close']]

    # Rename 'Adj Close' to ticker symbol
    df = df.rename(columns={'Adj Close': ticker})

    # Return the processed DataFrame and its row count
    return df, df.shape[0]

# Function to download and process data
def download_and_process_data(tickers, start, end, out_dir):
    """
    Download and process stock data.
    Args:
    - tickers (list): List of stock tickers.
    - start (str): Start date (YYYY-MM-DD).
    - end (str): End date (YYYY-MM-DD).
    - out_dir (str): Directory to save data.
    Returns:
    - pd.DataFrame: Combined data for tickers.
    - dict: Dictionary of ticker with data points before and after processing.
    """
    # Initialize empty DataFrame
    combined_df = None

    # Initialize a dictionary to store data points information
    data_points_info = {}

    # Loop through each ticker
    for ticker in tickers:
        # Fetch raw data
        raw_df, raw_count = fetch_data(ticker, start, end, out_dir)

        # Process data for current ticker
        processed_df, processed_count = process_data(raw_df,
                                                     ticker.split('.')[0])

        # Record the data points information
        data_points_info[ticker] = {
            'before_processing': raw_count,
            'after_processing': processed_count
        }

        # Merge processed data with combined_df
        combined_df = (processed_df if combined_df is None
                       else pd.merge(combined_df, processed_df,
                                     on='Date', how='inner'))

    # Return the combined DataFrame and the data points info
    return combined_df, data_points_info

# Define tickers for each dataset
tickers_train = ['RMV.L', 'SBRY.L', 'BATS.L', 'PSON.L','TW.L']
tickers_test = ['SGE.L', 'PHNX.L', 'SHEL.L', 'HIK.L', 'TSCO.L']
tickers_extra =['SMDS.L', 'ULVR.L', 'NG.L',  'GSK.L',
                'MNDI.L']

# Define date ranges for datasets
date_train = ('2014-01-02', '2023-12-29')
date_test = ('2024-01-02', '2024-03-28')
date_extra = ('2024-04-02', '2024-08-28')

# Directories for saving data
dirs = ('train_data_', 'test_data_', 'extra_data_')

# Download and process training data
combined_train_data, train_data_points = \
download_and_process_data(tickers_train, *date_train, dirs[0])

# Download and process test data
combined_test_data, test_data_points =\
 download_and_process_data(tickers_test, *date_test, dirs[1])

# Download and process extra data
combined_extra_data, extra_data_points = \
download_and_process_data(tickers_extra, *date_extra, dirs[2])

# Print first few rows of each dataset
print(combined_train_data.head())
print(combined_test_data.head())
print(combined_extra_data.head())

# Print the data points information
print("Training Data Points:", train_data_points)
print("Test Data Points:", test_data_points)
print("Extra Data Points:", extra_data_points)
```

```python
def calculate_log_returns(df):
    """
    Calculate the log returns for each ticker in the DataFrame.

    Args:
    - df (pd.DataFrame): DataFrame with 'Date' as the first column and
     adjusted close prices for each ticker.

    Returns:
    - pd.DataFrame: A new DataFrame containing log returns for each ticker,
    with the 'Date' column retained.
    """
    # Copy  original DataFrame
    log_returns = df.copy()

    # Loop through each column (excluding 'Date') to calculate log returns
    for column in df.columns[1:]:
        log_returns[column] = np.log(df[column] / df[column].shift(1))

    # Drop the first row
    log_returns = log_returns.dropna()

    # Return the DataFrame with calculated log returns
    return log_returns

def plot_log_returns(log_returns_data, title="Log Returns Time Series"):
    """
    Plot of log returns time series  in subplots with 3 plots per row.

    Args:
    - log_returns_data (pd.DataFrame): DataFrame containing log returns with
    'Date' as the first column.
    - title (str): The main title of the plot.

    Returns:
    - None. Displays the time series plots of log returns.
    """
    # Extract ticker symbols from the columns (excluding 'Date')
    tickers = log_returns_data.columns[1:]

    # Calculate the number of rows needed for subplots (3 plots per row)
    num_tickers = len(tickers)
    num_rows = (num_tickers + 2) // 3

    # Create a figure with subplots arranged in a grid
    fig, axs = plt.subplots(num_rows, 3, figsize=(18, num_rows * 5))

    # Set the main title for the entire figure
    fig.suptitle(title, fontsize=16)

    # Loop through each ticker to plot its log returns
    for i, ticker in enumerate(tickers):
        # Select the appropriate subplot
        ax = axs[i // 3, i % 3]

        # Plot the log returns against the Date
        ax.plot(log_returns_data['Date'], log_returns_data[ticker],
                label=ticker, linewidth=1.5)

        # Set the title, labels, and grid for each subplot
        ax.set_title(ticker, fontsize=12)
        ax.set_xlabel('Date')
        ax.set_ylabel('Log Return')
        ax.legend(loc='upper left', fontsize='small')
        ax.grid(True)

        # Rotate x-axis labels for better readability
        ax.tick_params(axis='x', rotation=45)

    # Remove any empty subplots
    for j in range(i + 1, num_rows * 3):
        fig.delaxes(axs.flatten()[j])

    # Adjust the layout to prevent overlapping
    plt.tight_layout(rect=[0, 0, 1, 0.95])

    # Display the plot
    plt.show()

# Function to plot box plots of adjusted close prices
def plot_box_plots(adj_close_data, title="Adjusted Close Price Box Plots",
                   currency="GBP", scale=""):
    """
    Create box plots for each ticker to detect outliers using the original
     adjusted close prices,
    with appropriate currency and scale.

    Args:
    - adj_close_data (pd.DataFrame): DataFrame containing adjusted close
     prices with 'Date' as the first column.
    - title (str): The main title of the plot.
    - currency (str): The currency unit of the adjusted close price ('GBP').
    - scale (str): The scale of the adjusted close price

    Returns:
    - None. Displays the box plots for adjusted close prices.
    """
    # Extract ticker symbols from the columns (excluding 'Date')
    tickers = adj_close_data.columns[1:]

    # Calculate the number of rows needed for subplots (3 plots per row)
    num_tickers = len(tickers)
    num_rows = (num_tickers + 2) // 3

    # Construct the y-axis label with currency and scale
    ylabel = f"Adjusted Close Price ({currency} {scale})".strip()

    # Create a figure with subplots arranged in a grid
    fig, axs = plt.subplots(num_rows, 3, figsize=(18, num_rows * 5))

    # Set the main title for the entire figure
    fig.suptitle(title, fontsize=16)

    # Loop through each ticker to create a box plot
    for i, ticker in enumerate(tickers):
        # Select the appropriate subplot
        ax = axs[i // 3, i % 3]

        # Create the box plot for the adjusted close prices
        sns.boxplot(y=adj_close_data[ticker], ax=ax)

        # Set the title and y-axis label for each subplot
        ax.set_title(ticker, fontsize=12)
        ax.set_ylabel(ylabel)

    # Remove any empty subplots
    for j in range(i + 1, num_rows * 3):
        fig.delaxes(axs.flatten()[j])

    # Adjust the layout to prevent overlapping
    plt.tight_layout(rect=[0, 0, 1, 0.95])

    # Display the plot
    plt.show()

# Calculate log returns for the training data
log_returns_train = calculate_log_returns(train_data)

# Plot the log returns time series for the training data
plot_log_returns(
    log_returns_train, title="Log Returns Time Series for Training Data")

# Plot box plots for the adjusted close prices in the training data
plot_box_plots(
    train_data, title= "Adjusted Close Price Box Plots for\n"
    f"Training Data",
               currency="GBP")

# Calculate log returns for the test data
log_returns_test = calculate_log_returns(test_data)

# Plot the log returns time series for the test data
plot_log_returns(
    log_returns_test, title="Log Returns Time Series for Test Data")

# Plot box plots for the adjusted close prices in the test data
plot_box_plots(
    test_data, title="Adjusted Close Price Box Plots for Test Data",
    currency="GBP")

# Calculate log returns for the extra data
log_returns_extra = calculate_log_returns(extra_data)

# Plot the log returns time series for the extra data
plot_log_returns(
    log_returns_extra, title="Log Returns Time Series for Extra Data")

# Plot box plots for the adjusted close prices in the extra data
plot_box_plots(
    extra_data, title="Adjusted Close Price Box Plots for Extra Data",
    currency="GBP")
```

# ## Overview

# This project aims to preprocess stock data by generating lagged features and calculating technical indicators,
# such as **EMA (Exponential Moving Average)** and **RSI (Relative Strength Index)**.
# These features are used for training machine learning models to predict future stock prices.

# ## Workflow

# 2. **Load Datasets**:

# 3. **Preprocess Data**:
#    - The stock data is preprocessed by setting the 'Date' column as the index, ensuring proper time-series handling.

#    - Technical indicators such as EMA and RSI are computed for each stock ticker.

#    - The data is scaled using `StandardScaler` to normalize the features and target variables.

#    - The preprocessed and scaled data is saved in CSV format for future model training.

# The steps ensure the data is clean, normalized, and properly structured, which helps the models to perform accurately
# and consistently when making predictions on future stock prices.


```python
# Set random seeds for reproducibility
set_random_seeds(seed_value=42)

def preprocess_data(train_data, test_data, extra_data, n_lags=2,
                    output_dir='processed_data'):
    """
    Preprocess stock data, create lagged features, and save processed data.
    Args:
    - train_data: DataFrame for training data.
    - test_data: DataFrame for test data.
    - extra_data: DataFrame for extra test data.
    - n_lags: Number of lagged features to create (fixed to 2 in this case).
    - output_dir: Directory to save the processed data.

    Returns:
    - Dictionary of preprocessed data for train, test, and extra datasets.
    """
    # Ensure output directory exists
    os.makedirs(output_dir, exist_ok=True)

    # Initialize dictionary for storing processed data
    preprocessed_data = {'train': {}, 'test': {}, 'extra': {}}

    # Datasets to iterate through
    datasets = {'train': train_data, 'test': test_data, 'extra': extra_data}

    feature_names_saved = False  # Track if feature names have been saved

    for name, data in datasets.items():
        print(f"Processing {name} data")

        # Ensure 'Date' is set as the index
        if data.index.name != 'Date':
            print(f"Indexing {name} by 'Date'")
            data['Date'] = pd.to_datetime(data['Date'])
            data.set_index('Date', inplace=True)

        for ticker in data.columns:
            print(f"Processing {ticker}")
            stock_data = data[[ticker]].copy()

            # Compute only EMA_12 and RSI indicators
            stock_data['EMA_12'] = ta.trend.ema_indicator(stock_data[ticker],
                                                          window=12)
            stock_data['RSI'] = ta.momentum.rsi(stock_data[ticker], window=14)

            # Create only lag 1 and lag 2 features
            for lag in range(1, n_lags + 1):
                stock_data[f'{ticker}_lag_{lag}'] = stock_data[ticker].shift(lag)

            # Drop any rows with missing values
            stock_data.dropna(inplace=True)

            if stock_data.empty:
                print(f"No data left for {ticker}")
                continue

            # Separate features and target
            X = stock_data.drop(columns=[ticker])
            y = stock_data[ticker]  # Target variable

            # Scale features and target using StandardScaler
            feature_scaler = StandardScaler()
            target_scaler = StandardScaler()

            X_scaled = feature_scaler.fit_transform(X)
            y_scaled = target_scaler.fit_transform(y.values.reshape(-1, 1)).flatten()

            # Convert scaled data back to DataFrame and Series
            X_scaled = pd.DataFrame(X_scaled, columns=X.columns, index=X.index)
            y_scaled = pd.Series(y_scaled, index=y.index)

            # Save feature names
            if name == 'train' and not feature_names_saved:
                feature_names = X.columns.tolist()  # Assuming X is a DataFrame
                joblib.dump(feature_names, 'xgb_feature_names.pkl')
                print("Feature names saved for future alignment.")
                feature_names_saved = True

            # Store scaled data in the preprocessed_data dictionary
            preprocessed_data[name][ticker] = {
                'X': X_scaled,
                'y': y_scaled,
                'feature_scaler': feature_scaler,
                'target_scaler': target_scaler
            }

            # Save processed data to CSV and pickle files
            X_scaled.to_csv(os.path.join(output_dir, f'{name}_{ticker}_X.csv'))
            y_scaled.to_csv(os.path.join(output_dir, f'{name}_{ticker}_y.csv'))
            joblib.dump(feature_scaler,
                        os.path.join(output_dir,
                                     f'{name}_{ticker}_feature_scaler.pkl'))
            joblib.dump(target_scaler,
                        os.path.join(output_dir,
                                     f'{name}_{ticker}_target_scaler.pkl'))

    # Print data shapes
    print("\nFinal shapes of preprocessed data:")
    for name, dataset in preprocessed_data.items():
        print(f"\n{name.capitalize()} data:")
        for ticker, data in dataset.items():
            print(f"  {ticker}: X shape {data['X'].shape},\n"
                  f"y shape {data['y'].shape}")

    return preprocessed_data


# Load datasets from CSV
train_data = pd.read_csv('output/train_data/train_data.csv')
test_data = pd.read_csv('output/test_data/test_data.csv')
extra_data = pd.read_csv('output/extra_data/extra_data.csv')

# Preprocess the datasets
preprocessed_data = preprocess_data(train_data, test_data, extra_data)
```

# ## Overview
# The function `combine_hist_train_datasets` is designed to combine preprocessed historical training data
# for multiple stock tickers into a single dataset. This combined dataset will include standardized feature names
# to ensure that the features are aligned correctly across different tickers.
#
# The main objective is to combine input features (X) and target values (y) from multiple stock tickers
# into a single matrix for model training. Additionally, the feature names are saved to ensure consistency
# during model training and inference.

# ## Workflow
# 1. **Initialize Empty Lists**:
#    - The function starts by initializing two empty lists: `combined_X` to hold the combined features
#      and `combined_y` to hold the combined target values.
#
# 2. **Define Standardized Feature Names**:
#    - A list of standardized feature names is defined, including indicators like `EMA_12`, `RSI`, and lagged features
#      (`lag_1`, `lag_2`).
#
# 3. **Loop through Each Ticker**:
#    - The function iterates over the training data of each ticker in the `preprocessed_data` dictionary.
#      For each ticker, the input features (X) and target values (y) are extracted and appended to the respective lists.
#
# 4. **Combine Feature and Target Data**:
#    - After looping through all tickers, the features from each ticker are stacked vertically using `np.vstack()`,
#      and the target values are concatenated into a single array.
#
# 5. **Save Feature Names**:
#    - If `save_feature_names` is set to `True`, the standardized feature names are saved to a file using `joblib.dump()`.
#
# 6. **Return Combined Data**:
#    - The function returns three items: the combined features (`combined_X`), combined target values (`combined_y`),
#      and the standardized feature names.

# ## Conclusion
# The `combine_hist_train_datasets` function effectively merges the preprocessed data from multiple stock tickers,
# standardizing the feature names across all the tickers. This combined dataset can now be used for training machine
# learning models, ensuring that all features are aligned and properly scaled. Additionally, by saving the feature names,
# the function ensures consistency during both training and inference phases, contributing to model accuracy and reliability.


```python
def combine_hist_train_datasets(preprocessed_data, save_feature_names=True):
    """
    Combine historical training datasets with standardized feature names.

    Args:
    - preprocessed_data: Dict with preprocessed data.
    - save_feature_names: Bool, whether to save the feature names to a file.

    Returns:
    - combined_X: Combined input features with standardized names.
    - combined_y: Combined target values.
    - feature_names: Standardized feature names for the combined dataset.
    """
    combined_X = []  # List to hold combined features
    combined_y = []  # List to hold combined targets

    # Standardized feature names based on the preprocessing step
    standardized_feature_names = ['EMA_12', 'RSI', 'lag_1', 'lag_2']

    # Loop through each ticker's training data
    for ticker, data in preprocessed_data['train'].items():
        # Append feature data
        combined_X.append(data['X'].values)
        # Append target data
        combined_y.append(data['y'].values)

    # Vertically stack features from all tickers
    combined_X = np.vstack(combined_X)
    # Concatenate target values into one array
    combined_y = np.concatenate(combined_y)

    #  save feature names to a file
    if save_feature_names:
        joblib.dump(standardized_feature_names, 'trained_feature_names.pkl')
        print(f"Feature names saved: {standardized_feature_names}")

    return combined_X, combined_y, standardized_feature_names


# Call the function
combined_X, combined_y, feature_names = combine_hist_train_datasets(preprocessed_data)

# Output the shapes of the combined data and the standardized feature names
print(f"Combined X shape: {combined_X.shape}")
print(f"Combined y shape: {combined_y.shape}")
print(f"Feature names: {feature_names}")
```

```python
# Set random seeds for reproducibility
set_random_seeds(seed_value=42)

# Define the reshape_for_lstm function
def reshape_for_lstm(X, sequence_length):
    """
    Reshape the input data for LSTM.

    Args:
    - X: Input feature data (2D array).
    - sequence_length: Number of timesteps to use.

    Returns:
    - X_reshaped: Reshaped input data for LSTM.

    The function creates sequences of data using a
    rolling window approach to match LSTM input shape.
    """
    X_reshaped = []  # List to hold reshaped sequences

    # Create sequences using a rolling window
    for i in range(len(X) - sequence_length + 1):
        X_reshaped.append(X[i:i + sequence_length])

    return np.array(X_reshaped)  # Convert to numpy array

# Custom RMSE metric
def rmse(y_true, y_pred):
    return K.sqrt(K.mean(K.square(y_pred - y_true)))

# Function to build and compile the LSTM model with Attention before BiLSTM
def build_lstm_model(input_shape):
    """
    Build and compile an LSTM model with Attention before the BiLSTM layer.
    """
    inputs = Input(shape=input_shape)

    # Set Attention mechanism before BiLSTM
    attention = Attention(name='attention_layer')([inputs, inputs])

    # BiLSTM with regularization
    x = Bidirectional(LSTM(50, return_sequences=True,
                           kernel_regularizer=l2(0.005)))(attention)

    # Set dropout to prevent overfitting
    x = Dropout(0.6)(x)

    # Set Flatten layer
    x = Flatten()(x)

    # Set Dense layer with ReLU and increased regularization
    x = Dense(32, activation='relu', kernel_regularizer=l2(0.005))(x)

    # Dropout in dense layer
    x = Dropout(0.6)(x)

    # Output layer for regression
    outputs = Dense(1)(x)

    # Compile model with Adam, RMSE, MAE
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(learning_rate=0.0001),
                  loss='mse', metrics=['mae', rmse])

    return model

# Define the sequence length
sequence_length = 30

# Reshape the combined training data for LSTM
X_train_reshaped = reshape_for_lstm(combined_X, sequence_length)
y_train_reshaped = combined_y[sequence_length - 1:]

# Confirm data shapes
print(f"X_train shape: {X_train_reshaped.shape}")
print(f"y_train shape: {y_train_reshaped.shape}")

# Split data into train and validation
split_ratio = 0.8
split_index = int(X_train_reshaped.shape[0] * split_ratio)
X_train, X_val = X_train_reshaped[:split_index], X_train_reshaped[split_index:]
y_train, y_val = y_train_reshaped[:split_index], y_train_reshaped[split_index:]

# Build the LSTM model with Attention before BiLSTM
model = build_lstm_model(X_train.shape[1:])

# Early stopping to prevent overfitting
early_stopping = EarlyStopping(monitor='val_loss', patience=5,
                               restore_best_weights=True)

# Train model with reduced epochs, validation, and early stopping
history = model.fit(X_train, y_train, epochs=50, batch_size=50,
                    validation_data=(X_val, y_val),
                    callbacks=[early_stopping])

# Save the trained model after training completes
model.save('lstm_model_with_attention_before_bilstm.keras')

# Plot training and validation loss
plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

# Plot training and validation MAE
plt.subplot(1, 2, 2)
plt.plot(history.history['mae'], label='Train MAE')
plt.plot(history.history['val_mae'], label='Val MAE')
plt.title('Model MAE')
plt.xlabel('Epoch')
plt.ylabel('MAE')
plt.legend()

plt.show()

# Display model summary
model.summary()

```

```python
# Set random seeds for reproducibility
set_random_seeds(seed_value=42)

# Define the custom RMSE metric
@tf.keras.utils.register_keras_serializable()
def rmse(y_true, y_pred):
    return tf.sqrt(tf.reduce_mean(tf.square(y_pred - y_true)))

# Function to build and compile the LSTM model with hyperparameters(Optuna)

def build_htlstm_model(input_shape, trial):
    """
    Build and compile an LSTM model with hyperparameters tuned by Optuna.

    Args:
    - input_shape: Shape of the input data (timesteps, features).
    - trial: Optuna trial object to suggest hyperparameters.

    Returns:
    - model: Compiled Keras LSTM model.
    """
    # Hyperparameter tuning using Optuna
    n_units = trial.suggest_int('units', 32, 128, step=32)
    dropout_rate = trial.suggest_float('dropout', 0.3, 0.6, step=0.1)
    learning_rate = trial.suggest_float('learning_rate', 1e-4, 1e-2, log=True)

    inputs = Input(shape=input_shape)

    # Set Attention mechanism before BiLSTM
    attention = Attention(name='attention_layer')([inputs, inputs])

    # BiLSTM with regularization and variable units
    x = Bidirectional(LSTM(n_units, return_sequences=True,
                           kernel_regularizer=tf.keras.regularizers.l2(
                               0.005)))(attention)

    # Set dropout to prevent overfitting with variable dropout rate
    x = Dropout(dropout_rate)(x)

    # Set Flatten layer
    x = Flatten()(x)

    # Set Dense layer with ReLU and increased regularization
    x = Dense(32, activation='relu',
              kernel_regularizer=tf.keras.regularizers.l2(0.005))(x)

    # Dropout in dense layer
    x = Dropout(dropout_rate)(x)

    # Output layer for regression
    outputs = Dense(1)(x)

    # Compile model with Adam optimizer and RMSE metric
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(learning_rate=learning_rate),
                  loss='mse', metrics=['mae', rmse])

    return model

# Define the Optuna objective function
def objective(trial):
    """
    Objective function for Optuna hyperparameter optimization.

    Args:
    - trial: An Optuna trial object used to sample hyperparameters.

    Returns:
    - val_rmse: Validation RMSE score.
    """
    model = build_htlstm_model(X_train.shape[1:], trial)

    # Early stopping
    early_stopping = EarlyStopping(monitor='val_loss', patience=5,
                                   restore_best_weights=True)

    # Define the checkpoint callback
    checkpoint = ModelCheckpoint('model_checkpoint.keras',
                                 monitor='val_loss',
                                 save_best_only=True,
                                 mode='min',
                                 verbose=1)

    # Train the model
    history = model.fit(X_train, y_train, epochs=50, batch_size=50,
                        validation_data=(X_val, y_val),
                        callbacks=[early_stopping, checkpoint], verbose=0)

    # Evaluate on validation data
    val_loss, val_mae, val_rmse = model.evaluate(X_val, y_val, verbose=0)

    return val_rmse

# Data preparation
sequence_length = 30  # Sequence length for LSTM
X_train_reshaped = reshape_for_lstm(combined_X, sequence_length)
y_train_reshaped = combined_y[sequence_length - 1:]

# Split data into train and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_train_reshaped,
                                                  y_train_reshaped,
                                                  test_size=0.2, random_state=42)

# Run Optuna to optimize hyperparameters
study = optuna.create_study(direction='minimize')
study.optimize(objective, n_trials=10)  # Limit to 10 trials for time efficiency

# Get the best hyperparameters from Optuna
best_trial = study.best_trial
print(f"Best hyperparameters: {best_trial.params}")

# Rebuild the model with the best hyperparameters
best_model = build_htlstm_model(X_train.shape[1:], best_trial)

# Train the final model using early stopping and the best hyperparameters
early_stopping = EarlyStopping(monitor='val_loss', patience=5,
                               restore_best_weights=True)

# Define the checkpoint callback
checkpoint = ModelCheckpoint('model_checkpoint.keras',
                             monitor='val_loss',
                             save_best_only=True,
                             save_weights_only=False,
                             mode='min',
                             verbose=1)

history = best_model.fit(X_train, y_train, epochs=50, batch_size=50,
                         validation_data=(X_val, y_val),
                         callbacks=[early_stopping, checkpoint])

# Save the trained model
best_model.save('best_htlstm_model.keras')

# Plot training and validation loss
plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

# Plot training and validation MAE
plt.subplot(1, 2, 2)
plt.plot(history.history['mae'], label='Train MAE')
plt.plot(history.history['val_mae'], label='Val MAE')
plt.title('Model MAE')
plt.xlabel('Epoch')
plt.ylabel('MAE')
plt.legend()

plt.show()

# Display model summary
best_model.summary()

# Visualize and save the model architecture
plot_model(best_model, to_file='best_lstm_model_architecture.png',
           show_shapes=True, show_layer_names=True)
```

# ## Title: Evaluating and Saving Performance Metrics of a Tuned BiLSTM Model

# ## Overview
# This section focuses on evaluating the performance of a tuned **Bidirectional LSTM (BiLSTM)** model using metrics such as **Mean Absolute Error (MAE)**, **Root Mean Squared Error (RMSE)**, and **R² score**. The evaluation is performed on both the **Test Data** and **Extra Test Data**, and the results are presented in tabular form. The metrics are visualized, saved to a **CSV file**, and exported as a **Word document** for further analysis.

# ## Workflow

# 1. **Set Random Seed**:
#    - A random seed is set for reproducibility using `set_random_seeds(seed_value=42)` to ensure that results are consistent across runs.

# 2. **Model Evaluation**:
#    - The function `evaluate_model` is used to calculate the **MAE**, **RMSE**, and **R²** for both the test and extra test datasets. These metrics evaluate the performance of the model in predicting stock prices. If the target values are scaled, the function also reverses the scaling to evaluate the model with actual values.

# 3. **Plot Evaluation Metrics**:
#    - The function `plot_metrics` is used to visualize the evaluation metrics (MAE, RMSE, R²) for each dataset. Bar charts are generated and saved as PNG files for easy interpretation of the results.

# 4. **Saving Evaluation Metrics in Word Table**:
#    - The function `save_as_word_table` converts the evaluation metrics into a **Word document** with a formatted table. This document can be shared or analyzed in a professional report format.

# 5. **Define Custom RMSE Metric**:
#    - A custom **RMSE** (Root Mean Squared Error) function is defined and registered with **Keras** to be used in the model evaluation.

# 6. **Load the Model with Custom RMSE**:
#    - The tuned **BiLSTM model** is loaded using the `load_model` function with the custom RMSE metric registered. This model will be evaluated using the test and extra test datasets.

# 7. **Evaluating the Model on Test and Extra Test Data**:
#    - The model is evaluated on both the **Test Data** and the **Extra Test Data** by reshaping the data for LSTM input and passing it through the model. The resulting predictions are compared with the actual target values to compute the evaluation metrics. These metrics are stored in a dictionary for further analysis.

# 8. **Saving and Displaying Results**:
#    - The evaluation results are saved in a **CSV file** and a **Word document** for further review and analysis. The results are also displayed directly in the notebook for immediate inspection.

# ## Conclusion
# This process evaluates the performance of the **tuned BiLSTM model** on two datasets using key regression metrics (MAE, RMSE, R²). The results are saved in multiple formats (CSV, Word) and visualized using bar charts, making the analysis clear and accessible. These steps are crucial for understanding the model's performance and identifying potential areas for improvement in future iterations.




```python
# Set random seeds for reproducibility
set_random_seeds(seed_value=42)

# Set the sequence length
sequence_length = 30

def evaluate_model(model, X, y, scaler=None):
    """
    Evaluate model performance.

    Args:
    - model: Trained LSTM model.
    - X: Input features.
    - y: True target values.
    - scaler: Scaler to reverse scaling.

    Returns:
    - Dictionary with MAE, RMSE, R².
    """
    predictions = model.predict(X)

    if scaler is not None:
        y = scaler.inverse_transform(y.reshape(-1, 1)).flatten()
        predictions = scaler.inverse_transform(predictions).flatten()

    mae = mean_absolute_error(y, predictions)
    rmse = np.sqrt(mean_squared_error(y, predictions))
    r2 = r2_score(y, predictions)

    return {"MAE": mae, "RMSE": rmse, "R²": r2}

def plot_metrics(metrics, title):
    """
    Plot MAE, RMSE, R² values.

    Args:
    - metrics: Evaluation metrics dict.
    - title: Title for the plot.

    Returns:
    - None. Displays the plot.
    """
    fig, ax = plt.subplots(1, 3, figsize=(15, 5))
    ax[0].bar(metrics.keys(), [m["MAE"] for m in metrics.values()])
    ax[0].set_title('MAE')
    ax[1].bar(metrics.keys(), [m["RMSE"] for m in metrics.values()])
    ax[1].set_title('RMSE')
    ax[2].bar(metrics.keys(), [m["R²"] for m in metrics.values()])
    ax[2].set_title('R²')
    fig.suptitle(title)
    plt.savefig(f"{title}_metrics.png")
    plt.show()

def save_as_word_table(df, filename="evaluation_metrics.docx"):
    """
    Save the DataFrame as a Word table.

    Args:
    - df: DataFrame containing the metrics.
    - filename: Name of the Word file to save.
    """
    from docx import Document
    doc = Document()
    doc.add_heading('Evaluation Metrics', 0)

    # Add table to document
    table = doc.add_table(rows=(df.shape[0] + 1), cols=df.shape[1] + 1)

    # Adding headers
    hdr_cells = table.rows[0].cells
    hdr_cells[0].text = "Dataset/Ticker"
    for i, column_name in enumerate(df.columns):
        hdr_cells[i + 1].text = column_name

    # Adding data
    for i, idx in enumerate(df.index):
        row_cells = table.rows[i + 1].cells
        row_cells[0].text = str(idx)
        for j, value in enumerate(df.loc[idx]):
            row_cells[j + 1].text = str(value)

    # Save the document
    doc.save(filename)

@register_keras_serializable()
def rmse(y_true, y_pred):
    return tf.sqrt(tf.reduce_mean(tf.square(y_pred - y_true)))

# Load the model with the custom RMSE metric
model = load_model('best_htlstm_model.keras', custom_objects={'rmse': rmse})

all_metrics = {}

# Evaluate on test dataset
print("=== Tuned BiLSTM Test Data ===")
test_metrics = {}
for ticker, data in preprocessed_data['test'].items():
    X_test_reshaped = reshape_for_lstm(data['X'].values, sequence_length)
    y_test_reshaped = data['y'].values[sequence_length - 1:]
    scaler = data.get('target_scaler')
    metrics = evaluate_model(model, X_test_reshaped, y_test_reshaped, scaler)
    test_metrics[ticker] = metrics
    print(f"{ticker}: {metrics}")
all_metrics["Tuned BiLSTM Test Data"] = test_metrics

# Evaluate on extra test dataset
print("\n=== Tuned BiLSTM Extra Test Data ===")
extra_metrics = {}
for ticker, data in preprocessed_data['extra'].items():
    X_extra_reshaped = reshape_for_lstm(data['X'].values, sequence_length)
    y_extra_reshaped = data['y'].values[sequence_length - 1:]
    scaler = data.get('target_scaler')
    metrics = evaluate_model(model, X_extra_reshaped, y_extra_reshaped, scaler)
    extra_metrics[ticker] = metrics
    print(f"{ticker}: {metrics}")
all_metrics["Tuned BiLSTM Extra Test Data"] = extra_metrics

# Create DataFrame for results
df = pd.DataFrame.from_dict(
    {(i, j): all_metrics[i][j]
     for i in all_metrics.keys()
     for j in all_metrics[i].keys()},
    orient='index')

# Save and display table
df.to_csv('tuned_BiLSTM_evaluation_metrics.csv')
save_as_word_table(df, "tuned_BiLSTM_evaluation_metrics.docx")
print(df)  # Display DataFrame directly

```

```python
# Set random seeds for reproducibility
set_random_seeds(seed_value=42)


def plot_actual_vs_predicted_subplots_with_dates(results, dataset_type,
                                                 tickers_per_row=2):
    """
    Plot actual vs predicted stock prices for multiple tickers using
    subplots with dates on the x-axis.

    Args:
    - results: Dictionary containing actual and predicted values for each ticker.
    - dataset_type: Whether the dataset is 'Test Data' or 'Extra Test Data'.
    - tickers_per_row: Number of subplots per row.

    Returns:
    - None. Displays and saves the plots with subplots.
    """
    num_tickers = len(results)
    num_rows = math.ceil(num_tickers / tickers_per_row)

    fig, axes = plt.subplots(num_rows, tickers_per_row,
                             figsize=(14, 5 * num_rows))
    axes = axes.flatten()  # Flatten the axes array for easier iteration

    for i, (ticker, (y_actual, y_pred, dates)) in enumerate(results.items()):
        axes[i].plot(dates, y_actual, label='Actual', color='blue',
                     linewidth=2, alpha=0.7)
        axes[i].plot(dates, y_pred, label='Predicted', color='red',
                     linestyle='--', linewidth=2, alpha=0.7)
        axes[i].set_title(f'{ticker}', fontsize=14)
        axes[i].set_xlabel('Date', fontsize=12)
        axes[i].set_ylabel('Stock Price', fontsize=12)

        # Format dates on x-axis
        axes[i].xaxis.set_major_locator(mdates.AutoDateLocator())
        axes[i].xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
        axes[i].grid(True, which='both', linestyle='--', linewidth=0.5)
        axes[i].legend()
        plt.setp(axes[i].xaxis.get_majorticklabels(), rotation=45, ha="right")

    # Adjust layout
    plt.suptitle(f'Actual vs Predicted Prices ({dataset_type})', fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.95])  # Adjust layout
    plt.savefig(f'{dataset_type}_actual_vs_predicted_with_dates.png', dpi=300)
    plt.show()

def evaluate_and_collect_results_with_dates(model, preprocessed_data,
                                            dataset_type, sequence_length=30):
    """
    Evaluate the model on the given dataset and collect actual vs predicted
     values with dates.

    Args:
    - model: Trained model (e.g., LSTM).
    - preprocessed_data: Preprocessed data dictionary containing features and
     targets.
    - dataset_type: Either 'test' or 'extra' to choose the dataset.
    - sequence_length: Sequence length for reshaping data (default: 30).

    Returns:
    - Dictionary of actual and predicted values with dates for each ticker.
    """
    results = {}
    for ticker, data in preprocessed_data[dataset_type].items():
        print(f"Processing {ticker}...")

        # Reshape data for LSTM
        X_reshaped = reshape_for_lstm(data['X'].values, sequence_length)
        y_reshaped = data['y'].values[sequence_length - 1:]
        scaler = data.get('target_scaler')

        # Get predictions
        predictions = model.predict(X_reshaped)
        y_pred = scaler.inverse_transform(predictions).flatten()
        y_actual = scaler.inverse_transform(y_reshaped.reshape(-1, 1)).flatten()

        # Extract dates (aligned with actual and predicted data)
        dates = data['X'].index[sequence_length - 1:]

        # Store the actual, predicted values and dates for plotting later
        results[ticker] = (y_actual, y_pred, dates)

    return results

# Load the trained model
model = load_model('best_htlstm_model.keras', custom_objects={'rmse': rmse})

# Collect actual vs predicted results for test data with dates
test_results_with_dates = evaluate_and_collect_results_with_dates(
    model, preprocessed_data, dataset_type='test', sequence_length=30)

# Plot subplots for test data with dates on x-axis
plot_actual_vs_predicted_subplots_with_dates(results=test_results_with_dates,
                                             dataset_type=\
                                             'Tuned BiLSTM Test Data')


# Plot subplots for extra test data with dates on x-axis
plot_actual_vs_predicted_subplots_with_dates(
    results=extra_results_with_dates, dataset_type=\
    'Tuned BiLSTM Extra Test Data')

```

# ## Title: Hyperparameter Optimization for XGBoost Using Particle Swarm Optimization (PSO)

# ## Overview
# This section demonstrates how to optimize an **XGBoost regression model** using **Particle Swarm Optimization (PSO)**. The PSO algorithm tunes the hyperparameters of the model to minimize the **validation RMSE**. Once the best hyperparameters are found, the final XGBoost model is trained on the full dataset, and its feature importance is visualized.

# ## Workflow

# 1. **Set Random Seeds for Reproducibility**:
#    - Random seeds are set to ensure that the results remain consistent across multiple runs.

# 2. **PSO Objective Function**:
#    - The `pso_objective` function is defined to minimize the validation **RMSE**. It accepts hyperparameters (such as learning rate, max depth, and others), trains the model on the training data, and evaluates it on the validation data. The objective function returns the validation **RMSE** as the metric to minimize during PSO optimization.

# 3. **Evaluate XGBoost Model Performance**:
#    - The `evaluate_xgb_model` function computes the model's performance on the test set using three key metrics:
#      - **MAE (Mean Absolute Error)**: Measures the average absolute difference between actual and predicted values.
#      - **RMSE (Root Mean Squared Error)**: Penalizes larger errors more heavily and provides insight into the model's overall prediction accuracy.
#      - **R² (R-squared score)**: Measures how well the model's predictions match the actual data.

# 4. **Train/Validation Data Split**:
#    - The data is split into **training** and **validation** sets using an 80/20 split to train the model and evaluate its performance on unseen data.

# 5. **PSO Hyperparameter Optimization**:
#    - **PSO bounds** are defined for the hyperparameters:
#      - `learning_rate`, `max_depth`, `n_estimators`, `subsample`, and `colsample_bytree`.
#    - The **PSO algorithm** is then run to minimize the validation **RMSE**, returning the best hyperparameters and the corresponding validation RMSE.

# 6. **Train the Final Model Using Best Hyperparameters**:
#    - The final XGBoost model is trained using the best hyperparameters returned by PSO on the full dataset (combined training and validation data).
#    - The trained model is saved to a **JSON** file (`xgb_final_model.json`).

# 7. **Feature Importance**:
#    - **Feature importance** is extracted from the trained XGBoost model using the "weight" importance type, which shows how often each feature was used in the model's decision trees.
#    - The feature importance is then visualized in a **bar chart**, where each feature is plotted along with its importance score.

# ## Conclusion
# Using **Particle Swarm Optimization (PSO)**, we optimized the hyperparameters of an **XGBoost regression model** to minimize the validation **RMSE**. After finding the best hyperparameters, the final model was trained on the full dataset and saved for future use. Additionally, we visualized the **feature importance** to understand which features contributed the most to the model’s predictions.


```python
# Set random seeds for reproducibility
set_random_seeds(seed_value=42)

def pso_objective(params):
    """
    Objective function for PSO to minimize validation RMSE.

    Args:
    - params: list of hyperparameters (learning_rate, max_depth,
     n_estimators, subsample, colsample_bytree)

    Returns:
    - val_rmse: Validation RMSE
    """
    learning_rate, max_depth, n_estimators, subsample, colsample_bytree = params
    max_depth = int(max_depth)
    n_estimators = int(n_estimators)

    xgb_params = {
        'objective': 'reg:squarederror',
        'learning_rate': learning_rate,
        'max_depth': max_depth,
        'n_estimators': n_estimators,
        'subsample': subsample,
        'colsample_bytree': colsample_bytree,
        'eval_metric': ['rmse', 'mae'],
        'seed': 42
    }

    # Train the XGBoost model
    dtrain = xgb.DMatrix(X_train, label=y_train)
    dval = xgb.DMatrix(X_val, label=y_val)
    evals_result = {}

    model = xgb.train(xgb_params, dtrain, num_boost_round=n_estimators,
                      evals=[(dtrain, 'train'), (dval, 'val')],
                      evals_result=evals_result,
                      early_stopping_rounds=10,
                      verbose_eval=False)

    val_rmse = min(evals_result['val']['rmse'])

    return val_rmse


def evaluate_xgb_model(model, X, y, scaler=None):
    """
    Evaluate XGBoost model performance.

    Args:
    - model: Trained XGBoost model.
    - X: Input features.
    - y: True target values.
    - scaler: Scaler to reverse scaling.

    Returns:
    - Dictionary with MAE, RMSE, R².
    """
    dtest = xgb.DMatrix(X)
    predictions = model.predict(dtest)

    if scaler is not None:
        y = scaler.inverse_transform(y.reshape(-1, 1)).flatten()
        predictions = scaler.inverse_transform(
            predictions.reshape(-1, 1)).flatten()

    mae = mean_absolute_error(y, predictions)
    rmse = np.sqrt(mean_squared_error(y, predictions))
    r2 = r2_score(y, predictions)

    return {"MAE": mae, "RMSE": rmse, "R²": r2}

# Split data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(combined_X,
                                                  combined_y, test_size=0.2,
                                                  random_state=42)

# Define PSO bounds for hyperparameters
lb = [0.01, 3, 50, 0.5, 0.5]  # lower bounds
ub = [0.3, 10, 300, 1.0, 1.0]  # upper bounds

# Run PSO optimization
best_params, best_rmse = pso(pso_objective, lb, ub, swarmsize=20, maxiter=50)

# Output the best hyperparameters and validation RMSE
print("Best Hyperparameters:", best_params)
print("Best Validation RMSE:", best_rmse)

# Train final XGBoost model with best hyperparameters on full training data
final_params = {
    'objective': 'reg:squarederror',
    'learning_rate': best_params[0],
    'max_depth': int(best_params[1]),
    'n_estimators': int(best_params[2]),
    'subsample': best_params[3],
    'colsample_bytree': best_params[4],
    'eval_metric': ['rmse', 'mae'],  # Track both RMSE and MAE
    'seed': 42
}

# Train the model on the full dataset
dtrain_full = xgb.DMatrix(combined_X, label=combined_y)
final_model = xgb.train(final_params, dtrain_full,
                        num_boost_round=int(best_params[2]))

# Save the final model to JSON
final_model.save_model('xgb_final_model.json')
print("Final model saved as 'xgb_final_model.json'.")


# Load feature names used during training
feature_names = joblib.load('xgb_feature_names.pkl')

# Get feature importance from the trained XGBoost model
importance = final_model.get_score(importance_type='weight')

# Convert feature importance into a sorted list
importance = {feature_names[int(k[1:])]: v for k, v in importance.items()}
sorted_importance = sorted(importance.items(), key=lambda item: item[1],
                           reverse=True)

# Plot the feature importance as a bar chart
plt.figure(figsize=(10, 6))
plt.bar([x[0] for x in sorted_importance], [x[1] for x in sorted_importance])
plt.xticks(rotation=45)
plt.title("Feature Importance")
plt.xlabel("Features")
plt.ylabel("Importance Score")
plt.tight_layout()
plt.show()
```

# ## Overview

```python
# Set random seeds for reproducibility
set_random_seeds(seed_value=42)

def combine_train_data(preprocessed_data):
    """
    Combine training data from multiple tickers into a single dataset.

    Args:
    - preprocessed_data (dict): Preprocessed data dictionary containing
                                feature (X) and target (y) values for each ticker.

    Returns:
    - combined_X (numpy array): Combined feature values from all tickers.
    - combined_y (numpy array): Combined target values from all tickers.
    - feature_names (list): Extracted feature names from the preprocessed data.
    """
    combined_X = []
    combined_y = []
    feature_names = None

    for ticker, data in preprocessed_data['train'].items():
        combined_X.append(data['X'])
        combined_y.append(data['y'])

        # Get feature names from the first ticker
        if feature_names is None:
            feature_names = data['X'].columns.tolist()

    return np.vstack(combined_X), np.concatenate(combined_y), feature_names

def plot_actual_vs_predicted(y_actual, y_pred, subset_size=500, window_size=10):
    """
    Plot actual vs predicted stock prices, with optional smoothing via
     rolling mean.

    Args:
    - y_actual (numpy array): Actual target values.
    - y_pred (numpy array): Predicted target values by the model.
    - subset_size (int): Number of data points to visualize in the plot
    (default=500).
    - window_size (int): Window size for rolling mean to smooth the plot
     (default=10).
    """
    # Select a subset of the data for clarity (first 500 points)
    y_actual_subset = pd.Series(y_actual[:subset_size])
    y_pred_subset = pd.Series(y_pred[:subset_size])

    # Apply rolling mean to smooth the data (smoothing window of 10 by default)
    y_actual_smoothed = y_actual_subset.rolling(window=window_size).mean()
    y_pred_smoothed = y_pred_subset.rolling(window=window_size).mean()

    plt.figure(figsize=(10, 6))

    # Plot the actual and predicted prices with transparency and thinner lines
    plt.plot(y_actual_smoothed, label='Actual Prices (Smoothed)', color='blue',
             alpha=0.6, linewidth=1)
    plt.plot(y_pred_smoothed, label='Predicted Prices (Smoothed)', color='red',
             alpha=0.6, linewidth=1)

    # Add title and labels
    plt.title('SVR Predicted vs Actual Stock Prices (Validation Set)',
              fontsize=14)
    plt.xlabel('Time', fontsize=12)
    plt.ylabel('Price', fontsize=12)

    # Add legend
    plt.legend()

    # Show grid
    plt.grid(True)

    # Display the plot
    plt.show()

def train_and_evaluate_svr(preprocessed_data, subset_size=500, test_size=0.2):
    """
    Train a Support Vector Regression (SVR) model and evaluate it using RMSE
     and MAE. Visualize the actual vs predicted prices.

    Args:
    - preprocessed_data (dict): Preprocessed data dictionary containing
                                feature (X) and target (y) values for each
                                ticker.
    - subset_size (int): Number of data points to visualize in the plot
    (default=500).
    - test_size (float): Proportion of the data to use as the validation set
     (default=0.2).
    """

    # Prepare training data and extract feature names
    X_train, y_train, feature_names = combine_train_data(preprocessed_data)

    # Save the feature names for later evaluation use
    joblib.dump(feature_names, 'svr_feature_names.pkl')

    # Split training data into train/validation sets
    X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
        X_train, y_train, test_size=test_size, random_state=42
    )

    # Feature scaling
    scaler = StandardScaler()  # Use StandardScaler as requested
    X_train_scaled = scaler.fit_transform(X_train_split)
    X_val_scaled = scaler.transform(X_val_split)

    # Save the scaler for later use during evaluation
    joblib.dump(scaler, 'svr_scaler.pkl')

    # Define the SVR model with fixed hyperparameters
    svr_model = SVR(kernel='rbf', C=100, gamma=0.5)

    # Train the SVR model
    svr_model.fit(X_train_scaled, y_train_split)

    # Make predictions on the validation set
    y_val_pred = svr_model.predict(X_val_scaled)

    # Calculate RMSE for validation
    rmse = np.sqrt(mean_squared_error(y_val_split, y_val_pred))
    print(f"Validation RMSE: {rmse}")

    # Calculate MAE for evaluation
    mae = mean_absolute_error(y_val_split, y_val_pred)
    print(f"Evaluation MAE: {mae}")

    # Plot actual vs. predicted prices for validation set
    plot_actual_vs_predicted(y_val_split, y_val_pred, subset_size=subset_size,
                             window_size=10)

    # Use permutation importance to estimate feature importance
    result = permutation_importance(svr_model, X_val_scaled, y_val_split,
                                    n_repeats=10, random_state=42)

    # Extract importance scores and sort them by importance
    sorted_idx = result.importances_mean.argsort()

    # Plot feature importance using the extracted feature names
    plt.figure(figsize=(10, 6))
    plt.barh(range(len(sorted_idx)), result.importances_mean[sorted_idx],
             align="center")
    plt.yticks(range(len(sorted_idx)), [feature_names[i] for i in sorted_idx])
    plt.title("Permutation Feature Importance (SVR)")
    plt.xlabel("Importance Score")
    plt.ylabel("Features")
    plt.tight_layout()
    plt.show()

    # Save the trained SVR model
    joblib.dump(svr_model, 'svr_final_model.pkl')  # Save the model
    print("SVR model saved as 'svr_final_model.pkl'")

# Call the function to train and evaluate the SVR model
train_and_evaluate_svr(preprocessed_data, subset_size=500, test_size=0.2)
```

```python
# Set random seeds for reproducibility
set_random_seeds(seed_value=42)

def combine_train_data(preprocessed_data):
    """
    Combine training data from multiple tickers into a single dataset.

    Args:
    - preprocessed_data (dict): Preprocessed data dictionary containing
                                feature (X) and target (y) values for each ticker.

    Returns:
    - combined_X (numpy array): Combined feature values from all tickers.
    - combined_y (numpy array): Combined target values from all tickers.
    """
    combined_X = []
    combined_y = []
    for ticker, data in preprocessed_data['train'].items():
        combined_X.append(data['X'])
        combined_y.append(data['y'])
    return np.vstack(combined_X), np.concatenate(combined_y)

def plot_actual_vs_predicted(y_actual, y_pred, subset_size=500, window_size=10):
    """
    Plot actual vs predicted stock prices, with optional smoothing via
    rolling mean.

    Args:
    - y_actual (numpy array): Actual target values.
    - y_pred (numpy array): Predicted target values by the model.
    - subset_size (int): Number of data points to visualize in the plot
     (default=500).
    - window_size (int): Window size for rolling mean to smooth the plot
     (default=10).
    """
    y_actual_subset = pd.Series(y_actual[:subset_size])
    y_pred_subset = pd.Series(y_pred[:subset_size])

    y_actual_smoothed = y_actual_subset.rolling(window=window_size).mean()
    y_pred_smoothed = y_pred_subset.rolling(window=window_size).mean()

    plt.figure(figsize=(10, 6))
    plt.plot(y_actual_smoothed, label='Actual Prices (Smoothed)',
             color='blue', alpha=0.6, linewidth=1)
    plt.plot(y_pred_smoothed, label='Predicted Prices (Smoothed)',
             color='red', alpha=0.6, linewidth=1)
    plt.title('SVR Predicted vs Actual Stock Prices (Validation Set)',
              fontsize=14)
    plt.xlabel('Time', fontsize=12)
    plt.ylabel('Price', fontsize=12)
    plt.legend()
    plt.grid(True)
    plt.show()

def train_and_evaluate_svr_with_gridsearch(preprocessed_data, subset_size=500,
                                           test_size=0.2):
    """
    Train and optimize an SVR model using Grid Search and evaluate it using
    RMSE and MAE. Visualize the actual vs predicted prices.

    Args:
    - preprocessed_data (dict): Preprocessed data dictionary containing
                                feature (X) and target (y) values for each
                                 ticker.
    - subset_size (int): Number of data points to visualize in the plot
    (default=500).
    - test_size (float): Proportion of the data to use as the validation set
     (default=0.2).
    """

    # Prepare training data
    X_train, y_train = combine_train_data(preprocessed_data)

    # Split training data into train/validation sets
    from sklearn.model_selection import train_test_split
    X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
        X_train, y_train, test_size=test_size, random_state=42
    )

    # Feature scaling using StandardScaler
    from sklearn.preprocessing import StandardScaler
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train_split)
    X_val_scaled = scaler.transform(X_val_split)

    # Save  scaler for consistency in future evaluations
    joblib.dump(scaler, 'svr_final_model.pkl')

    # Define hyperparameter grid for Grid Search
    param_grid = {
        'C': [0.1, 1, 10],  # Experiment with fewer values for efficiency
        'gamma': [0.01, 0.1, 1],
        'kernel': ['rbf']  # Only RBF kernel
    }

    # Initialize SVR model
    from sklearn.svm import SVR
    svr = SVR()

    # Set up GridSearchCV
    from sklearn.model_selection import GridSearchCV
    grid_search = GridSearchCV(
        svr, param_grid, cv=3, scoring='neg_mean_squared_error',
        verbose=1, n_jobs=-1
    )

    # Train using grid search
    grid_search.fit(X_train_scaled, y_train_split)

    # Get best model from grid search
    best_svr_model = grid_search.best_estimator_

    # Evaluate the best model on validation data
    y_val_pred = best_svr_model.predict(X_val_scaled)

    # Calculate RMSE and MAE for validation
    from sklearn.metrics import mean_squared_error, mean_absolute_error
    rmse = np.sqrt(mean_squared_error(y_val_split, y_val_pred))
    mae = mean_absolute_error(y_val_split, y_val_pred)

    print(f"Best Parameters from Grid Search: {grid_search.best_params_}")
    print(f"Validation RMSE: {rmse}")
    print(f"Evaluation MAE: {mae}")

    # Plot actual vs. predicted prices for validation set
    plot_actual_vs_predicted(y_val_split, y_val_pred, subset_size=subset_size,
                             window_size=10)

    # Return best model
    return best_svr_model

# Call the function to train and evaluate the SVR model using Grid Search
best_svr_model = train_and_evaluate_svr_with_gridsearch(preprocessed_data,
                                                        subset_size=500,
                                                        test_size=0.2)

# Save the best SVR model
joblib.dump(best_svr_model, 'best_svr_model.pkl')  # Save the model
print("Best SVR model saved as 'best_svr_model.pkl'")
```

# ## Title: Evaluating a Tuned SVR Model on Test and Extra Test Data

# ## Overview
# In this section, we evaluate the performance of a **tuned Support Vector Regression (SVR)** model on both **Test Data** and **Extra Test Data**. The code processes the input data, aligns features to match the trained model's format, and evaluates the model’s performance using key regression metrics such as **MAE (Mean Absolute Error)**, **RMSE (Root Mean Squared Error)**, and **R² (R-squared score)**. Results are stored in a DataFrame and saved for further analysis.

# ## Workflow

# 2. **Align Features in Test/Extra Data**:
#    - The `align_features` function is used to align the features in the test and extra test datasets to match the feature names used during training. It handles missing features by filling them with **NaN** values, which are later replaced by zeros or mean imputation.

# 3. **Evaluate the SVR Model**:
#    - The `evaluate_svr_model` function evaluates the trained **SVR model** on the given test data. It calculates **MAE**, **RMSE**, and **R²** metrics to measure the model's performance.
#    - If the target values were scaled during training, the function reverses the scaling to ensure that the evaluation is done on the original scale of the stock prices.

# 4. **Load the Tuned SVR Model**:
#    - The trained **SVR model** is loaded using **`joblib`**, and the feature names used during training are also loaded. This ensures that the input features in the test dataset are properly aligned before evaluation.

# 5. **Evaluate the Model on Test Data**:
#    - The model is evaluated on the **Test Data** if available in the preprocessed dataset. For each ticker in the test set, the true and predicted values are compared, and the evaluation metrics (**MAE, RMSE, R²**) are computed and printed.

# 6. **Evaluate the Model on Extra Test Data**:
#    - If available, the model is also evaluated on the **Extra Test Data**, which provides additional insight into how well the model generalizes to unseen data.
#    - The same evaluation process is followed as for the Test Data, and the results are printed for each ticker in the extra test set.

# 7. **Store and Save the Results**:
#    - If metrics are computed, they are stored in a **DataFrame** and saved as a CSV file for future analysis. The results allow for easy comparison of the model’s performance across different datasets and tickers.

# ## Conclusion
# The tuned **SVR model** is evaluated on both **Test Data** and **Extra Test Data**, and key performance metrics (**MAE, RMSE, R²**) are computed for each stock ticker. These metrics provide insight into how well the model predicts stock prices, and the results are saved for further analysis. This step ensures that the model’s performance is thoroughly evaluated and documented, aiding in decision-making and future improvements.


```python
# Set random seeds for reproducibility
set_random_seeds(seed_value=42)


def align_features(X, expected_columns):
    """
    Align features in test/extra data to match the feature names used
    during training.

    Args:
    - X: Input features for test/extra data.
    - expected_columns: The feature names used during training.

    Returns:
    - X_aligned: Aligned features with missing columns filled as 0 and
     NaNs imputed.
    """
    X_df = pd.DataFrame(X, columns=expected_columns)
    X_aligned = X_df.reindex(columns=expected_columns, fill_value=np.nan)

    # Handle missing values by filling NaNs with 0
    X_aligned.fillna(0, inplace=True)

    # Handle any remaining NaN values using mean imputation
    imputer = SimpleImputer(strategy='mean')
    X_imputed = imputer.fit_transform(X_aligned)

    return X_imputed


# Function to evaluate the SVR model
def evaluate_svr_model(model, X, y, target_scaler_path=None):
    """
    Evaluate the SVR model and return metrics.

    Args:
    - model: Trained SVR model.
    - X: Input features for evaluation.
    - y: True target values (as a numpy array).
    - target_scaler_path: Path to the target scaler for inverse
     transformation (optional).

    Returns:
    - Dictionary with MAE, RMSE, and R² metrics.
    """
    # Convert y to NumPy array (if it's a pandas Series)
    if isinstance(y, pd.Series):
        y = y.values

    # Make predictions
    predictions = model.predict(X)

    if target_scaler_path:
        # Inverse scaling for both true and predicted values
        target_scaler = joblib.load(target_scaler_path)
        y_true_original = target_scaler.inverse_transform(
            y.reshape(-1, 1)).flatten()
        predictions_original = target_scaler.inverse_transform(
            predictions.reshape(-1, 1)).flatten()
    else:
        y_true_original = y
        predictions_original = predictions

    # Calculate evaluation metrics
    mae = mean_absolute_error(y_true_original, predictions_original)
    rmse = np.sqrt(mean_squared_error(y_true_original, predictions_original))
    r2 = r2_score(y_true_original, predictions_original)

    return {"MAE": mae, "RMSE": rmse, "R²": r2}


# Load the saved SVR model
svr_model = joblib.load('best_svr_model.pkl')

# Load feature names from training
feature_names = joblib.load('xgb_feature_names.pkl')

# Initialize metrics container
all_metrics = {}

# Check if 'test' data is available in preprocessed_data
if 'test' in preprocessed_data:
    print("=== Base SVR Test Data ===")
    test_metrics = {}
    for ticker, data in preprocessed_data['test'].items():
        X_test_reshaped = align_features(data['X'], feature_names)
        y_test_reshaped = data['y']
        scaler_path = f'processed_data/test_{ticker}_target_scaler.pkl'
        metrics = evaluate_svr_model(svr_model, X_test_reshaped,
                                     y_test_reshaped, scaler_path)
        test_metrics[ticker] = metrics
        print(f"{ticker}: {metrics}")

    all_metrics["Tuned  SVR Test Data"] = test_metrics
else:
    print("No test data found in preprocessed_data.")

# Check if 'extra' test data is available in preprocessed_data
if 'extra' in preprocessed_data:
    print("\n=== Tuned  SVR Extra Test Data ===")
    extra_metrics = {}
    for ticker, data in preprocessed_data['extra'].items():
        X_extra_reshaped = align_features(data['X'], feature_names)
        y_extra_reshaped = data['y']
        scaler_path = f'processed_data/extra_{ticker}_target_scaler.pkl'
        metrics = evaluate_svr_model(svr_model, X_extra_reshaped,
                                     y_extra_reshaped, scaler_path)
        extra_metrics[ticker] = metrics
        print(f"{ticker}: {metrics}")

    all_metrics["Tuned SVR Extra Test Data"] = extra_metrics
else:
    print("No extra test data found in preprocessed_data.")

# Create DataFrame for results if there are any metrics
if all_metrics:
    df = pd.DataFrame.from_dict(
        {(dataset, ticker): all_metrics[dataset][ticker]
         for dataset in all_metrics.keys()
         for ticker in all_metrics[dataset].keys()},
        orient='index')

    # Save and display table
    df.to_csv('TUned SVR evaluation_metrics_svr.csv')
    print(df)
else:
    print("No metrics were computed as there was no test or extra data available.")
```

# ## Title: Evaluating Feature Importance for an SVR Model Using Permutation Importance

# ## Overview
# This section focuses on evaluating the **feature importance** of an **SVR (Support Vector Regression)** model using **Permutation Importance**. Permutation Importance helps identify which features most impact the model’s predictions by randomly shuffling feature values and measuring how much the prediction error changes. This method is especially useful for models like SVR, which do not inherently provide feature importance.

# ## Workflow

# 1. **Load the Pre-trained SVR Model**:
#    - The **SVR model** is loaded from a **pickle** file (`best_svr_model.pkl`). This model has already been trained on the stock price data and is ready for evaluation.

# 2. **Load Feature Names**:
#    - The feature names used during training are loaded from the file `xgb_feature_names.pkl`. These feature names will be used to ensure that the test data is aligned properly and to label the feature importance plot.

# 3. **Align Features in Test Data**:
#    - The test data is aligned with the training features using the `align_features()` function. This ensures that the test dataset contains the same features as the training set, and any missing features are handled appropriately.

# 4. **Compute Permutation Feature Importance**:
#    - The **permutation_importance** function from **`sklearn.inspection`** is used to compute the importance of each feature. It works by shuffling each feature and measuring the drop in the model's prediction performance. The more the model's performance degrades, the more important that feature is considered.
#    - The importance scores are averaged over multiple repetitions (`n_repeats=10`) to ensure robustness.

# 5. **Sort and Plot Feature Importance**:
#    - The computed feature importance values are sorted in descending order.
#    - A **horizontal bar plot** is generated to visualize the feature importance, where the x-axis represents the **Permutation Importance Score** and the y-axis shows the **feature names**.
#    - This plot provides a clear interpretation of which features were most important to the SVR model in making predictions.

# ## Conclusion
# By using **Permutation Importance**, we can interpret the **SVR model's** predictions and identify which features had the greatest impact on its performance. This analysis provides valuable insights into the model's decision-making process, especially for complex datasets like stock price prediction, where understanding feature importance can guide further improvements to the model or feature engineering.
